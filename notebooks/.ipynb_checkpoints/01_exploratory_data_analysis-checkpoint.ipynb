```python
import numpy as np
import pandas as pd
from typing import Tuple


def prepare_xgb_data(
    df: pd.DataFrame,
    train_start: str,
    train_end: str,
    test_start: str,
    test_end: str
) -> Tuple[np.ndarray, np.ndarray]:
    
    df = df.copy()

    X_train = df.loc[train_start:train_end].drop(columns=['Label']).values
    y_train = df.loc[train_start:train_end]['Label'].values

    X_test = df.loc[test_start:test_end].drop(columns=['Label']).values
    y_test = df.loc[test_start:test_end]['Label'].values

    assert X_train.shape[0] == y_train.shape[0], 'X_train and y_train have a different number of rows'
    assert X_test.shape[0] == y_test.shape[0], 'X_test and y_test have a different number of rows'
    assert X_train.shape[1] == X_test.shape[1], 'X_train and X_test have a different number of features'

    return X_train, y_train, X_test, y_test

X_train, y_train, X_test, y_test = prepare_xgb_data(
    test, 
    train_start='2010-01-01',
    train_end = '2019-12-31',
    test_start = '2020-01-01',
    test_end = '2025-12-31')
```

```python
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import log_loss, accuracy_score, confusion_matrix
import time

results_benchmark = {}

for i, t in enumerate((1, 5, 21)):
    X_train, y_train, X_test, y_test = prepare_xgb_data(
        pd.read_parquet(f'../data/processed/processed_{t}.parquet'),
        train_start='2010-01-01',
        train_end = '2019-12-31',
        test_start = '2020-01-01',
        test_end = '2025-12-31')
    
    model = XGBClassifier(eval_metric='logloss', random_state=42)
    
    param_grid = {'model__n_estimators': [35, 40, 45],      
                  'model__max_depth': [2, 3, 4],              
                  'model__learning_rate': [0.05, 0.1, 0.15], 
                  'model__subsample': [0.15, 0.2, 0.25],    
                  'model__colsample_bytree': [0.8, 1], 
                  'model__gamma': [6.25, 6.5, 6.75],
                  'model__reg_alpha': [0, 0.1, 0.5],
                  'model__reg_lambda': [1, 1.5, 2]}
    
    pipe = Pipeline(
        steps=[
            ("scaler", StandardScaler()),
            ("model", model)
        ]
    )

    print(f'Starting GridSearch for XGBoost Classifier for hold days {t}.\n')
    start = time.perf_counter()
    
    grid_search = GridSearchCV(estimator=pipe,
                               param_grid=param_grid,
                               cv=TimeSeriesSplit(n_splits=5),                 
                               scoring='accuracy',                 
                               verbose=0,
                               n_jobs=-1)
    
    # Fit
    grid_result = grid_search.fit(X_train, y_train)
    
    # Predict
    results_benchmark[f'Prediction_{t}'] = grid_search.best_estimator_.predict_proba(X_test)[:,1]
    
    end = time.perf_counter()

    print(f'GridSearch complete. Time taken = {end - start:.2f}s.\n') 
    
    if i<2:
        print('Beginning next loop.\n')
```
